```python
import sympy as sp
sp.init_printing()
```

```python
t,m1,m2,g,l = sp.symbols("t m_1 m_2 g l")
r = sp.Function("r")(t)
theta = sp.Function("theta")(t)
```

```python
r_dot = sp.diff(r,t)
theta_dot = sp.diff(theta,t)
r_2dot = sp.diff(r_dot,t)
theta_2dot = sp.diff(theta_dot,t)
```

```python
T = (1/2)*(m1+m2)*r_dot**2 + (1/2)*m1*r**2*theta_dot**2
T
```

```python
V = m1*g*r*(1-sp.cos(theta)) - m2*g*(l-r)
V
```

```python
# langrangian
L = T - V
L
```

```python
dLdrdot = sp.diff(L,r_dot)
dLdrdotdt = sp.diff(dLdrdot,t)
dLdr = sp.diff(L,r)
EL_r = dLdrdotdt - dLdr
EL_r
```

```python
dLdthetadot = sp.diff(L,theta_dot)
dLdthetadotdt = sp.diff(dLdthetadot,t)
dLdtheta = sp.diff(L,theta)
EL_theta = dLdthetadotdt - dLdtheta
EL_theta
```

```python
sp.solve(EL_r,r_2dot)
```

```python
sp.solve(EL_theta,theta_2dot)
```
